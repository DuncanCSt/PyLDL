## create data set train and testing folds

```python
from pathlib import Path
from pyldl.utils import load_dataset
import os

dataset_dir = Path("../dataset")
datasets = sorted([p.name for p in dataset_dir.iterdir()]) if dataset_dir.exists() else []

print(datasets)

for dataset in datasets:
    X, D = load_dataset(dataset.removesuffix('.mat'))
    print(f"{dataset}: {X.shape} {D.shape}")
```

```python
from sklearn.model_selection import KFold
import numpy as np
import scipy.io as sio

datasets_to_use = ['Movie', 'Natural_Scene', 'SBU_3DFE', 'SJAFFE', 'emotion6', 'Twitter', 'Flickr', 'Yeast_spo5']

for dataset in datasets_to_use:
    path = dataset
    # If subfolder exists, skip, else create it
    if os.path.exists(path):
        print(f"Directory {path} already exists, skipping...")
    else:
        os.makedirs(path)
        print(f"Directory {path} created.")

    data = sio.loadmat(dataset_dir / f"{dataset}.mat")

    kf = KFold(n_splits=10, shuffle=True, random_state=42)
    for fold, (train_index, test_index) in enumerate(kf.split(data['features'])):
        X_train, X_test = data['features'][train_index], data['features'][test_index]
        D_train, D_test = data['labels'][train_index], data['labels'][test_index]

        train_file = os.path.join(path, f"{fold}_train_{dataset}.mat")
        test_file = os.path.join(path, f"{fold}_test_{dataset}.mat")

        sio.savemat(train_file, {"features": X_train, "labels": D_train})
        sio.savemat(test_file, {"features": X_test, "labels": D_test})
        
```
